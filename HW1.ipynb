## Task 1

1. Load the file. Report the number of rows, the number of columns, and how many
   values of `days_kept` are missing.


```python
# Reading the file is given -- pandas turns the blank days_kept fields into
# NaN automatically, which is what you want.
loans = pd.read_csv(os.path.join(FOLDER, f"loans_{STUDENT_NO}.csv"))
loans.head()
```

```python
n_rows = loans.shape[0]           # TODO: how many rows does the file have?
n_cols = loans.shape[1]           # TODO: how many columns does the file have?
n_missing_days = loans["days_kept"].isna().sum()   # TODO: how many values of days_kept are missing?
print(n_rows, n_cols, n_missing_days)
```

## Task 2

2. Report which `member_type` is the most common and its share of all loans,
   rounded to three decimals.


```python
counts = loans["member_type"].value_counts()
counts
```

```python
top_member_type = counts.idxmax()         # TODO: name of the most common member_type
top_member_type_share = round(counts.max()/len(loans), 3) # TODO: its share of ALL loans (not just the non-missing ones), rounded to 3 decimals
print(top_member_type, top_member_type_share)
```

## Task 3

3. Report which faculty has the longest mean `days_kept`, and that mean, rounded
   to two decimals. Missing values are excluded, not treated as zero.


```python
# groupby(...).mean() already skips NaN, so missing loans are excluded rather
# than treated as zero -- which is what the task asks for.
fac_mean = loans.groupby("faculty")["days_kept"].mean()
fac_mean
```

```python
faculty_longest = fac_mean.idxmax()           # TODO: which faculty has the longest mean days_kept?
faculty_longest_mean_days = round(fac_mean.max(),2)  # TODO: that mean, rounded to 2 decimals
print(faculty_longest, faculty_longest_mean_days)
```

## Task 4

4. Report the late-return rate for undergraduates and for staff, each rounded to
   three decimals.


```python
# returned_late is a boolean column, so its mean IS the rate.
late = loans.groupby("member_type")["returned_late"].mean()
late
```

```python
late_rate_undergraduate = round(late["undergraduate"],3)   # TODO: rounded to 3 decimals
late_rate_staff = round(late["staff"],3)          # TODO: rounded to 3 decimals
print(late_rate_undergraduate, late_rate_staff)
```
